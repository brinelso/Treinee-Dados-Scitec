# 🏆 Desafio Final - Sistema SciTec

Este notebook verifica valores de entrada e identifica possíveis alertas no sistema.

```python

valores_brutos: list[int] = [10, 50, 150, 20, 300]

def verificar_sistema(valores: list[int]) -> None:
    alertas: list[int] = []
    
    for valor in valores:
        if valor > 100:
            alertas.append(valor)
    
    if len(alertas) == 0:
        print("Sistema Seguro")
    else:
        print(f"⚠️ Revisar valores: {alertas}")


verificar_sistema(valores_brutos)
```
