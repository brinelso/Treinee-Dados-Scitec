# 3. Otimizar números inteiros

```python
df['year'] = df['year'].astype('int16')
df['month'] = df['month'].astype('int8')
df['victim_count'] = df['victim_count'].astype('int16')
```

```python
df.memory_usage(deep=True).sum() / 1024**2
```

# 4. Converter colunas de texto para category

```python
colunas_categoria = [
    'day_of_week',
    'season',
    'time_of_day',
    'country',
    'area_type',
    'crime_type',
    'victim_gender',
    'victim_age_group',
    'weapon_used',
    'suspect_status',
    'case_status',
    'reporting_method',
    'lighting_condition',
    'crime_resolved'
]
    
for col in colunas_categoria:
    df[col] = df[col].astype('category')
```

```python
df.memory_usage(deep=True).sum() / 1024**2
```

# 5. Otimizar números decimais

```python
df['financial_loss_usd'] = df['financial_loss_usd'].astype('float32')
```

```python
df.memory_usage(deep=True).sum() / 1024**2
```

# 6. Criando outro arquivo otimizado

```python
df.to_csv('crime_rate_safety_analysis_optimized.csv', index=False)
```
